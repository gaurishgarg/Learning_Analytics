```python
import numpy as np
import pandas as pd
```

```python
interaction_df = pd.read_csv("robo.csv")
interaction_df.head()
```

```python
element_lookup = (
    interaction_df[
        ["elementId", "element", "elementType"]
    ]
    .drop_duplicates(subset=["elementId"])
)
```

```python
element_lookup.shape
```

```python
element_lookup
```

```python
interaction_clean = interaction_df[
    [
        "user",
        "timeStamp",
        "verb",
        "element",
        "elementId",
        "elementType",
        "duration",
        "attempted"
    ]
].copy()

interaction_clean = interaction_clean.rename(
    columns={
        "user": "user_id",
        "timeStamp": "interaction_timestamp"
    }
)

interaction_clean["interaction"] = (
    interaction_clean["verb"].astype("string")
    + " — "
    + interaction_clean["element"].astype("string")
)

interaction_clean.head()
```

```python
interaction_clean["interaction_timestamp"] = pd.to_datetime(
    interaction_clean["interaction_timestamp"],
    utc=True,
    errors="coerce"
)
```

```python
interaction_clean = interaction_clean[
    interaction_clean["element"].notna()
].copy()
```

```python
interaction_clean[
    ["user_id", "interaction_timestamp", "verb", "element", "elementId",
     "elementType", "interaction"]
].head()
```

```python
interaction_clean.to_csv("interaction_clean_final.csv", index=False)
```
