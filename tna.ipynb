```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


interacted = pd.read_csv("weights_total.csv", index_col=0)
gazed = pd.read_csv("weights_gaze.csv", index_col=0)
non_gaze = pd.read_csv("weights_non_gaze.csv", index_col=0)

for matrix in [interacted, gazed, non_gaze]:
    matrix.index = ["Cluster 0", "Cluster 1", "Cluster 2"]
    matrix.columns = ["Cluster 0", "Cluster 1", "Cluster 2"]


plt.figure(figsize=(7, 6))

sns.heatmap(
    interacted,
    annot=True,
    fmt=".3f",
    vmin=0,
    vmax=1,
    square=True,
    cbar_kws={"label": "Transition weight"}
)

plt.xlabel("To")
plt.ylabel("From")
plt.title("Transition Weights: Interacted")
plt.tight_layout()
plt.show()


plt.figure(figsize=(7, 6))

sns.heatmap(
    gazed,
    annot=True,
    fmt=".3f",
    vmin=0,
    vmax=1,
    square=True,
    cbar_kws={"label": "Transition weight"}
)

plt.xlabel("To")
plt.ylabel("From")
plt.title("Transition Weights: Gazed")
plt.tight_layout()
plt.show()


plt.figure(figsize=(7, 6))

sns.heatmap(
    non_gaze,
    annot=True,
    fmt=".3f",
    vmin=0,
    vmax=1,
    square=True,
    cbar_kws={"label": "Transition weight"}
)

plt.xlabel("To")
plt.ylabel("From")
plt.title("Transition Weights: Non-Gaze Interaction")
plt.tight_layout()
plt.show()
```
